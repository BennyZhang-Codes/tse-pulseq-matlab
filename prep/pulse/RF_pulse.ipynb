```python
import os
import numpy as np
import matplotlib.pyplot as plt
import sigpy.mri.rf as sigpyrf
import scipy.io as sio
```

### 1. refocusing pulse

```python
N = 512
d1 = 0.01
d2 = 0.01
ptype = 'se'
ftype = 'ls'
tbs = [4, 5, 6, 7, 8, 9, 10, 11, 12]
rfs = np.zeros((len(tbs), N), dtype=np.complex128)
for i in range(len(tbs)):
    rfs[i, :] = sigpyrf.slr.dzrf(N, tbs[i], ptype, ftype, d1, d2, True)

sio.savemat(f'SLR_refoc_{ftype}_{d1}_{d2}.mat', {'pulses': rfs, 'tbs': tbs})
```

### 2. inversion pulse

```python
N = 512
d1 = 0.01
d2 = 0.01
ptype = 'inv'
ftype = 'ls'
tbs = [4, 5, 6, 7, 8, 9, 10, 11, 12]
rfs = np.zeros((len(tbs), N), dtype=np.complex128)
for i in range(len(tbs)):
    rfs[i, :] = sigpyrf.slr.dzrf(N, tbs[i], ptype, ftype, d1, d2, True)

sio.savemat(f'SLR_inver_{ftype}_{d1}_{d2}.mat', {'pulses': rfs, 'tbs': tbs})
```

### 3. gSlider pulse

```python
G = 5
N = 512
d1 = 0.01
d2 = 0.01
tbs = [12, 14, 16, 18, 20]
rfs = np.zeros((len(tbs), G, N), dtype=np.complex128)
for i in range(len(tbs)):
    pulses = sigpyrf.slr.dz_gslider_rf(n=N, g=G, flip=np.pi/2 , phi=np.pi, tb=tbs[i], d1=d1, d2=d2, cancel_alpha_phs=True)
    rfs[i, ...] = np.transpose(pulses)

sio.savemat(f'gSlider_excit_{d1}_{d2}.mat', {'pulses': rfs, 'tbs': tbs})
```
